```python
print('promo_id là unique:', df['promo_id'].is_unique)
print('Số dòng trùng toàn bộ:', df.duplicated().sum())
```

```python
df['start_date'] = pd.to_datetime(df['start_date'])
df['end_date'] = pd.to_datetime(df['end_date'])
print('Số start_date không đọc được:', df['start_date'].isna().sum())
print('Số end_date không đọc được:', df['end_date'].isna().sum())
```

```python


# Đảm bảo các giá trị nan (cả dạng chuỗi và dạng thực) được chuyển về 'All Categories'
df['applicable_category'] = df['applicable_category'].replace(['nan', np.nan, 'NaN', 'None'], 'All Categories')

# Kiểm tra lại kết quả
print("Bảng thống kê category sau khi cập nhật:")
display(df['applicable_category'].value_counts(dropna=False))
```

```python
str_columns = ['promo_id', 'promo_name', 'promo_type', 'applicable_category', 'promo_channel']
for col in str_columns:
    df[col] = df[col].astype(str).str.strip()
```

```python
df = df.drop_duplicates(subset=['promo_id'])
assert (df['start_date'] <= df['end_date']).all(), "Phát hiện start_date lớn hơn end_date!"
```

```python
print('Tổng số dòng     :', len(df))
print('Tổng số dòng trùng promo_id:', df.duplicated(subset=['promo_id']).sum())
print('Tổng số null     :', df.isna().sum().sum())
# null là vì theo data dictionary applicable_category để trống tức là áp dụng cho mọi ngành hàng
```

```python
df.to_csv('promotions_silver.csv', index=False, na_rep='')
```
