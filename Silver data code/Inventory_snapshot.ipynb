```python
import pandas as pd
import numpy as np
```

```python
# Đọc file inventory.csv từ thư mục hiện tại
df = pd.read_csv("inventory.csv")

print("5 DÒNG ĐẦU TIÊN FILE INVENTORY")
df.head()
```

```python
# Kiểm tra kiểu dữ liệu & tổng số dòng/cột (xem có cột nào bị thiếu dữ liệu không)
print("THÔNG TIN CÁC CỘT & KIỂU DỮ LIỆU")
df.info()
```

```python
# 4. Thống kê nhanh các chỉ số cơ bản (Min, Max, Trung bình) của các cột dạng số
print("\n THỐNG KÊ MÔ TẢ CÁC CỘT SỐ")
display(df.describe())
```

```python
print("SỐ LƯỢNG GIÁ TRỊ NULL TRÊN MỖI CỘT:")
print(df.isna().sum())

# Kiểm tra trùng lặp khóa chính liên hợp (snapshot_date, product_id)
pk_cols = ["snapshot_date", "product_id"]
print("\nSố dòng trùng hoàn toàn 100%:", df.duplicated().sum())
print("Số dòng trùng khóa chính (snapshot_date, product_id):", df.duplicated(subset=pk_cols).sum())

```

```python
# 1. Kiểm tra giá trị các cột cờ (Flags) - Chỉ được phép chứa 0 hoặc 1
print("Các giá trị thực tế trong stockout_flag :", df["stockout_flag"].unique())
print("Các giá trị thực tế trong overstock_flag:", df["overstock_flag"].unique())
print("Các giá trị thực tế trong reorder_flag  :", df["reorder_flag"].unique())

# 2. Bắt lỗi giá trị âm hoặc bất thường
print("\n--- KIỂM TRA LỖI DỮ LIỆU BẤT THƯỜNG ---")
print("Số dòng có tồn kho (stock_on_hand) < 0  :", (df["stock_on_hand"] < 0).sum())
print("Số dòng có hàng bán (units_sold) < 0     :", (df["units_sold"] < 0).sum())
print("Số dòng có hàng nhập (units_received) < 0 :", (df["units_received"] < 0).sum())

fill_rate_error = ((df["fill_rate"] < 0) | (df["fill_rate"] > 1)).sum()
print("Số dòng có fill_rate nằm ngoài khoảng [0, 1]:", fill_rate_error)
```

```python
# 1. Chuyển snapshot_date sang dạng datetime
df["snapshot_date"] = pd.to_datetime(df["snapshot_date"], errors="coerce")

# 2. ĐỌC BẢNG PRODUCTS VÀ THỰC HIỆN FK CHECK
try:
    df_products = pd.read_csv("products.csv")
    valid_product_ids = set(df_products["product_id"].dropna().unique())
    
    initial_count = len(df)
    df = df[df["product_id"].isin(valid_product_ids)].copy()
    dropped_rows = initial_count - len(df)
    
    print(f"--- KẾT QUẢ FK CHECK (PRODUCT) ---")
    print(f"Tổng số dòng ban đầu : {initial_count}")
    print(f"Số dòng hợp lệ       : {len(df)}")
    print(f"Số dòng bị loại bỏ   : {dropped_rows}")

except FileNotFoundError:
    print("--- CẢNH BÁO: Không tìm thấy file products.csv, giữ nguyên dữ liệu ---")

# 3. Ép kiểu dữ liệu tối ưu bộ nhớ
df["product_id"] = df["product_id"].astype("int32")
df["stock_on_hand"] = df["stock_on_hand"].astype("int32")
df["units_received"] = df["units_received"].astype("int32")
df["units_sold"] = df["units_sold"].astype("int32")
df["stockout_days"] = df["stockout_days"].astype("int8")

for col in ["stockout_flag", "overstock_flag", "reorder_flag"]:
    df[col] = df[col].astype("int8")

# 4. Lọc đúng 12 cột chuẩn 3NF theo Lược đồ 3NF (Loại bỏ product_name, category, segment, year, month)
cols_3nf = [
    "snapshot_date", "product_id", "stock_on_hand", "units_received", 
    "units_sold", "stockout_flag", "fill_rate", "sell_through_rate", 
    "overstock_flag", "reorder_flag", "stockout_days", "days_of_supply"
]
df = df[cols_3nf]

# 5. Sắp xếp dữ liệu theo Khóa chính (snapshot_date, product_id)
df = df.sort_values(["snapshot_date", "product_id"]).reset_index(drop=True)

# 6. Kiểm tra lại thông tin DataFrame chuẩn 3NF
print("\n--- THÔNG TIN DATAFRAME SAU KHI CHUẨN HÓA 3NF & FK CHECK ---")
df.info()
```

```python
# 1. Định nghĩa đường dẫn file xuất trực tiếp
output_path = "Inventory_snapshot.csv"

# 2. Xuất dữ liệu ra file CSV
df.to_csv("Inventory_snapshot.csv", index=False, sep=",", encoding="utf-8-sig")

# 3. In thông báo hoàn tất
print(f"Tổng số dòng    : {len(df)}")
print(f"Tổng số cột     : {len(df.columns)}")
print(f"Đã xuất file thành công chuẩn 3NF tại: {output_path}")
```
