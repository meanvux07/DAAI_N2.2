## 🛠️ BƯỚC 1: Cấu hình Môi trường & Tự động Quét Tìm / Giải Nén File

```python
import os
import sys
import glob
import zipfile
import shutil
import pandas as pd
import numpy as np

# Cấu hình pandas
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', lambda x: '%.2f' % x)

# Thư mục xuất dữ liệu Gold
OUTPUT_DIR = 'Gold_Dimensions'
os.makedirs(OUTPUT_DIR, exist_ok=True)

# 1. Tự động giải nén zip nếu người dùng upload zip lên Colab
search_roots = ['.', 'Silver data', '/content', '/content/drive/MyDrive']
for root in search_roots:
    if os.path.exists(root):
        for zip_path in glob.glob(os.path.join(root, '*.zip')):
            if 'Gold' not in os.path.basename(zip_path):
                print(f"📦 Phát hiện file nén: [{zip_path}]. Đang tự động giải nén...")
                try:
                    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
                        zip_ref.extractall('.')
                    print(f"✓ Giải nén thành công [{zip_path}]!")
                except Exception as e:
                    print(f"⚠️ Không thể giải nén {zip_path}: {e}")

# 2. Hàm tìm kiếm file đệ quy thông minh
def find_and_read_csv(filename, required=True):
    target_name = os.path.basename(filename).lower()
    candidate_files = []
    for search_dir in ['.', 'Silver data', '/content', '..']:
        if os.path.exists(search_dir):
            for root, dirs, files in os.walk(search_dir):
                if OUTPUT_DIR in root or '.git' in root or '.ipynb_checkpoints' in root:
                    continue
                for f in files:
                    if f.lower() == target_name:
                        candidate_files.append(os.path.join(root, f))

    if not candidate_files:
        if required:
            raise FileNotFoundError(f"❌ KHÔNG TÌM THẤY FILE '{filename}'! Vui lòng kéo thả file vào Colab.")
        return None

    found_path = candidate_files[0]
    try:
        df = pd.read_csv(found_path, sep=',', low_memory=False)
    except Exception:
        df = pd.read_csv(found_path, sep=';', low_memory=False)

    df.columns = df.columns.str.replace('\ufeff', '').str.strip()
    print(f"✓ Đã đọc [{filename}] từ: [{found_path}] -> ({df.shape[0]:,} dòng, {df.shape[1]} cột)")
    return df

print("\n🔥 HỆ THỐNG ĐÃ SẴN SÀNG XÂY DỰNG 6 BẢNG DIMENSION CHUẨN!")
```

## 📅 BƯỚC 2: Xử lý và Xây dựng Bảng `Dim_Date` (Calendar Dimension - PK: `Date_Key`)

```python
print("=== 1. XÂY DỰNG BẢNG DIM_DATE ===")
df_orders = find_and_read_csv('orders.csv')

df_orders['order_date_dt'] = pd.to_datetime(df_orders['order_date'], errors='coerce')
valid_dates = df_orders['order_date_dt'].dropna()

min_date = valid_dates.min().floor('D')
max_date = valid_dates.max().ceil('D') + pd.Timedelta(days=365)
print(f"  -> Dải ngày giao dịch: từ {min_date.date()} đến {max_date.date()}")

date_series = pd.date_range(start=min_date, end=max_date, freq='D')
dim_date = pd.DataFrame({'Date_Timestamp': date_series})

dim_date['Date_Key'] = dim_date['Date_Timestamp'].dt.strftime('%Y%m%d').astype(int)
dim_date['Full_Date'] = dim_date['Date_Timestamp'].dt.strftime('%Y-%m-%d')
dim_date['Day_Of_Month'] = dim_date['Date_Timestamp'].dt.day.astype(int)
dim_date['Month_Number'] = dim_date['Date_Timestamp'].dt.month.astype(int)
dim_date['Month_Name'] = dim_date['Date_Timestamp'].dt.strftime('%B')
dim_date['Quarter_Number'] = dim_date['Date_Timestamp'].dt.quarter.astype(int)
dim_date['Year_Number'] = dim_date['Date_Timestamp'].dt.year.astype(int)
dim_date['Day_Of_Week'] = (dim_date['Date_Timestamp'].dt.dayofweek + 1).astype(int)
dim_date['Day_Name'] = dim_date['Date_Timestamp'].dt.strftime('%A')
dim_date['Is_Weekend'] = dim_date['Day_Of_Week'].isin([6, 7]).astype(int)

cols_date = ['Date_Key', 'Full_Date', 'Day_Of_Month', 'Month_Number', 'Month_Name',
             'Quarter_Number', 'Year_Number', 'Day_Of_Week', 'Day_Name', 'Is_Weekend']
dim_date = dim_date[cols_date]

print(f"✓ Xây dựng Dim_Date hoàn tất! Tổng số dòng: {len(dim_date):,}")
dim_date.head()
```

## 🌍 BƯỚC 3: Xử lý và Xây dựng Bảng `Dim_Geography` (Location Dimension - PK: `Zip_Code`)

```python
print("=== 2. XÂY DỰNG BẢNG DIM_GEOGRAPHY ===")
df_geo = find_and_read_csv('geography.csv')

df_geo['zip'] = df_geo['zip'].astype(str).str.strip()
for col in ['city', 'district', 'region']:
    if col in df_geo.columns:
        df_geo[col] = df_geo[col].astype(str).str.strip().replace({'nan': None, 'None': None, '': None})

dim_geo = df_geo.drop_duplicates(subset=['zip']).copy()
dim_geo = dim_geo.rename(columns={
    'zip': 'Zip_Code',
    'city': 'City',
    'district': 'District',
    'region': 'Region'
})

dim_geo.reset_index(drop=True, inplace=True)
dim_geo = dim_geo[['Zip_Code', 'City', 'District', 'Region']]

print(f"✓ Xây dựng Dim_Geography hoàn tất! Tổng số vùng: {len(dim_geo):,}")
dim_geo.head()
```

## 👤 BƯỚC 4: Xử lý và Xây dựng Bảng `Dim_Customer` (Customer Dimension - PK: `Customer_ID`)

```python
print("=== 3. XÂY DỰNG BẢNG DIM_CUSTOMER ===")
df_customers = find_and_read_csv('customers_silver.csv')

df_customers = df_customers.dropna(subset=['customer_id']).copy()
df_customers['customer_id'] = pd.to_numeric(df_customers['customer_id'], errors='coerce').astype(int)
df_customers['zip'] = df_customers['zip'].astype(str).str.strip()

valid_zips = set(dim_geo['Zip_Code'])
df_customers['Zip_Code'] = np.where(df_customers['zip'].isin(valid_zips), df_customers['zip'], None)

df_customers['signup_date_dt'] = pd.to_datetime(df_customers['signup_date'], errors='coerce')
df_customers['Signup_Date'] = df_customers['signup_date_dt'].dt.strftime('%Y-%m-%d')

df_customers['gender'] = df_customers['gender'].astype(str).str.strip().str.lower().replace({'nan': None, 'none': None, '': None})
df_customers['age_group'] = df_customers['age_group'].astype(str).str.strip().replace({'nan': None, 'none': None, '': None})
df_customers['acquisition_channel'] = df_customers['acquisition_channel'].astype(str).str.strip().replace({'nan': None, 'none': None, '': None})

dim_cust = df_customers.drop_duplicates(subset=['customer_id']).copy()
dim_cust = dim_cust.rename(columns={
    'customer_id': 'Customer_ID',
    'gender': 'Gender',
    'age_group': 'Age_Group',
    'acquisition_channel': 'Acquisition_Channel'
})

dim_cust.reset_index(drop=True, inplace=True)
dim_cust = dim_cust[['Customer_ID', 'Zip_Code', 'Signup_Date', 'Gender', 'Age_Group', 'Acquisition_Channel']]

print(f"✓ Xây dựng Dim_Customer hoàn tất! Tổng số khách hàng: {len(dim_cust):,}")
dim_cust.head()
```

## 🛍️ BƯỚC 5: Xử lý và Xây dựng Bảng `Dim_Product` (Product Dimension - PK: `Product_ID`)

```python
print("=== 4. XÂY DỰNG BẢNG DIM_PRODUCT ===")
df_product = find_and_read_csv('silver_product.csv')

df_product = df_product.dropna(subset=['product_id']).copy()
df_product['product_id'] = pd.to_numeric(df_product['product_id'], errors='coerce').astype(int)

for price_col in ['price', 'cogs']:
    if price_col in df_product.columns:
        df_product[price_col] = pd.to_numeric(
            df_product[price_col].astype(str).str.replace(',', '.'),
            errors='coerce'
        ).fillna(0.0).round(2)

for s_col in ['product_name', 'category', 'segment', 'size', 'color']:
    if s_col in df_product.columns:
        df_product[s_col] = df_product[s_col].astype(str).str.strip().replace({'nan': None, 'None': None, '': None})

dim_prod = df_product.drop_duplicates(subset=['product_id']).copy()
dim_prod = dim_prod.rename(columns={
    'product_id': 'Product_ID',
    'product_name': 'Product_Name',
    'category': 'Category',
    'segment': 'Segment',
    'size': 'Size',
    'color': 'Color',
    'price': 'Standard_Price',
    'cogs': 'Standard_Cost'
})

dim_prod.reset_index(drop=True, inplace=True)
dim_prod = dim_prod[['Product_ID', 'Product_Name', 'Category', 'Segment', 'Size', 'Color', 'Standard_Price', 'Standard_Cost']]

print(f"✓ Xây dựng Dim_Product hoàn tất! Tổng số sản phẩm: {len(dim_prod):,}")
dim_prod.head()
```

## 🏷️ BƯỚC 6: Xử lý và Xây dựng Bảng `Dim_Promotion` (Promotion Dimension - PK: `Promo_ID`)

```python
print("=== 5. XÂY DỰNG BẢNG DIM_PROMOTION ===")
df_promo = find_and_read_csv('promotions.csv', required=False)

if df_promo is not None:
    df_promo['promo_id'] = df_promo['promo_id'].astype(str).str.strip()
    df_promo = df_promo[df_promo['promo_id'] != ''].copy()

    if 'discount_value' in df_promo.columns:
        df_promo['Discount_Value'] = pd.to_numeric(
            df_promo['discount_value'].astype(str).str.replace(',', '.'),
            errors='coerce'
        ).fillna(0.0).round(2)

    for date_fld in ['start_date', 'end_date']:
        if date_fld in df_promo.columns:
            target_name = 'Start_Date' if date_fld == 'start_date' else 'End_Date'
            df_promo[target_name] = pd.to_datetime(df_promo[date_fld], errors='coerce').dt.strftime('%Y-%m-%d')

    for str_fld in ['promo_name', 'promo_type', 'applicable_category', 'promo_channel']:
        if str_fld in df_promo.columns:
            df_promo[str_fld] = df_promo[str_fld].astype(str).str.strip().replace({'nan': None, 'None': None, '': None})

    dim_promo = df_promo.drop_duplicates(subset=['promo_id']).copy()
    dim_promo = dim_promo.rename(columns={
        'promo_id': 'Promo_ID',
        'promo_name': 'Promo_Name',
        'promo_type': 'Promo_Type',
        'applicable_category': 'Applicable_Category',
        'promo_channel': 'Promo_Channel'
    })

    dim_promo.reset_index(drop=True, inplace=True)
    dim_promo = dim_promo[['Promo_ID', 'Promo_Name', 'Promo_Type', 'Discount_Value', 'Start_Date', 'End_Date', 'Applicable_Category', 'Promo_Channel']]
else:
    dim_promo = pd.DataFrame(columns=['Promo_ID', 'Promo_Name', 'Promo_Type', 'Discount_Value', 'Start_Date', 'End_Date', 'Applicable_Category', 'Promo_Channel'])

print(f"✓ Xây dựng Dim_Promotion hoàn tất! Tổng số khuyến mãi: {len(dim_promo):,}")
dim_promo.head()
```

## 👔 BƯỚC 7: Xử lý và Xây dựng Bảng `Dim_Sales_Employee` (Sales Employee Dimension - PK: `Sales_Employee_ID`)
* **Nguồn:** `sale_employee.csv`
* **Khóa chính (PK):** `Sales_Employee_ID` (Đã loại bỏ cột `Employee_Key`)
* **Danh mục thuộc tính:** `Sales_Employee_ID`, `Sales_Employee_Name`, `Marital_Status`, `Education_Level`, `Years_Experience`

```python
print("=== 6. XÂY DỰNG BẢNG DIM_SALES_EMPLOYEE ===")
df_emp = find_and_read_csv('sale_employee.csv', required=False)

if df_emp is not None:
    # 1. Làm sạch ID
    df_emp['sales_employee_id'] = df_emp['sales_employee_id'].astype(str).str.strip()
    df_emp = df_emp[df_emp['sales_employee_id'] != ''].copy()

    # 2. Làm sạch chuỗi văn bản
    for col in ['sales_employee_name', 'marital_status', 'education_level']:
        if col in df_emp.columns:
            df_emp[col] = df_emp[col].astype(str).str.strip().replace({'nan': None, 'None': None, '': None})

    # 3. Làm sạch số năm kinh nghiệm
    if 'years_experience' in df_emp.columns:
        df_emp['years_experience'] = pd.to_numeric(df_emp['years_experience'], errors='coerce').fillna(0).astype(int)

    # 4. Khử trùng lặp theo ID
    dim_emp = df_emp.drop_duplicates(subset=['sales_employee_id']).copy()

    # 5. Đổi tên cột chuẩn Gold (PK: Sales_Employee_ID)
    dim_emp = dim_emp.rename(columns={
        'sales_employee_id': 'Sales_Employee_ID',
        'sales_employee_name': 'Sales_Employee_Name',
        'marital_status': 'Marital_Status',
        'education_level': 'Education_Level',
        'years_experience': 'Years_Experience'
    })

    dim_emp.reset_index(drop=True, inplace=True)
    cols_emp = ['Sales_Employee_ID', 'Sales_Employee_Name', 'Marital_Status', 'Education_Level', 'Years_Experience']
    dim_emp = dim_emp[cols_emp]
else:
    dim_emp = pd.DataFrame(columns=['Sales_Employee_ID', 'Sales_Employee_Name', 'Marital_Status', 'Education_Level', 'Years_Experience'])

print(f"✓ Xây dựng Dim_Sales_Employee hoàn tất! Tổng số nhân viên: {len(dim_emp):,}")
dim_emp.head()
```

## 📊 BƯỚC 8: Kiểm tra Chất lượng Dữ liệu & Tự động Nén Zip Tải Về

```python
print("=================================================================")
print("          BÁO CÁO KIỂM TRA CHẤT LƯỢNG 6 BẢNG DIMENSIONS          ")
print("=================================================================\n")

dims_dict = {
    'Dim_Date': (dim_date, 'Date_Key'),
    'Dim_Geography': (dim_geo, 'Zip_Code'),
    'Dim_Customer': (dim_cust, 'Customer_ID'),
    'Dim_Product': (dim_prod, 'Product_ID'),
    'Dim_Promotion': (dim_promo, 'Promo_ID'),
    'Dim_Sales_Employee': (dim_emp, 'Sales_Employee_ID')
}

for name, (df_dim, pk_col) in dims_dict.items():
    pk_nulls = df_dim[pk_col].isnull().sum()
    pk_dups = df_dim[pk_col].duplicated().sum()

    status_pk = "PASSED ✓" if (pk_nulls == 0 and pk_dups == 0) else f"FAILED ✗ ({pk_dups} dups, {pk_nulls} nulls)"

    print(f"📦 Bảng [{name}]:")
    print(f"   • Tổng số bản ghi     : {len(df_dim):,}")
    print(f"   • Số lượng thuộc tính : {len(df_dim.columns)}")
    print(f"   • Khóa chính (PK: {pk_col}): {status_pk}")

    # Xuất file CSV UTF-8 có BOM
    output_file = os.path.join(OUTPUT_DIR, f"{name}.csv")
    df_dim.to_csv(output_file, index=False, encoding='utf-8-sig')
    print(f"   • Đã lưu CSV          : '{output_file}'\n")

# Nén toàn bộ thư mục kết quả
zip_filename = 'Gold_Dimensions'
shutil.make_archive(zip_filename, 'zip', OUTPUT_DIR)
print(f"✓ Đã nén toàn bộ 6 bảng Dimension thành file: '{zip_filename}.zip'")

# Tự động tải file zip về máy nếu đang trên Colab
try:
    from google.colab import files
    print("\n-> Đang tự động tải file 'Gold_Dimensions.zip' về máy tính của bạn...")
    files.download(f"{zip_filename}.zip")
except ImportError:
    print(f"\n-> Môi trường Local: Toàn bộ file đã được lưu tại thư mục '{OUTPUT_DIR}/' và '{zip_filename}.zip'")

print("\n=================================================================")
print("🎉 HOÀN THÀNH ETL TOÀN BỘ 6 BẢNG DIMENSION (NATURAL KEYS)!      ")
print("=================================================================")
```
